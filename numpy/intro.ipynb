```python
import numpy as np

marks=np.array([10,20,30,40,50])

bonus=5

new_marks=bonus+marks
print(new_marks)
```

```python
a=[2,3,4,5]

a_np=np.array(a)
b_np=a_np*2

print(a_np)
print(b_np)
```

```python
arr=[[1,2,3],
     [4,5,6],
     [7,8,9]]
print(arr)
print(type(arr))

n_arr=np.array(arr)
print(n_arr)
print(type(n_arr))
```
